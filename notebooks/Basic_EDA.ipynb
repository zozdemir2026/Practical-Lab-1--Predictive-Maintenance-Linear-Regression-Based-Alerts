# Basic Exploratory Data Analysis
Explore the robot performance dataset to understand its structure and values.

## Load Dataset
Get the CSV file path from the `Initialization` class and load the dataset into a pandas DataFrame.

```python
import sys
import os

sys.path.append(os.path.abspath(".."))

from src.initialization import Initialization
import pandas as pd

init = Initialization()
file_path = init.get_data_path()

df = pd.read_csv(file_path)
df.head()
```

## Check Missing Values
Check each column for missing data, and cleaning missing data column. Likewise time format changing from string to datetime format. 

```python
df.info()
# Only drop axis columns that actually exist in the dataframe
axis_cols = [
    "Axis #9", "Axis #10", "Axis #11", "Axis #12",
    "Axis #13", "Axis #14"
]
existing_axis_cols = [col for col in axis_cols if col in df.columns]

if existing_axis_cols:
    df = df.drop(columns=existing_axis_cols)

df["Time"] = pd.to_datetime(df["Time"], utc=True)
df.info()
```

### Save Cleaned Dataset
Save the cleaned dataset as a new CSV file for later use.

```python
import os

# Create the processed folder if it does not exist
os.makedirs("../data/processed", exist_ok=True)

# Save the cleaned dataframe
df.to_csv("../data/processed/RMBR4-2_cleaned.csv", index=False)

print("Cleaned dataset saved successfully.")
```
